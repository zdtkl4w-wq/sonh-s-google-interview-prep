```python
from google.colab import drive
drive.mount('/content/drive')

# ▼ neurofee.zip 을 올린 위치에 맞게 수정하세요
ZIP_PATH = '/content/drive/MyDrive/neurofee.zip'

import os, zipfile, shutil
shutil.rmtree('/content/neurofee', ignore_errors=True)
with zipfile.ZipFile(ZIP_PATH) as z:
    z.extractall('/content')

# zip 안에 neurofee/neurofee 처럼 한 단계 더 들어가 있는 경우도 자동 처리
root = '/content/neurofee'
if not os.path.exists(f'{root}/configs') and os.path.exists(f'{root}/neurofee/configs'):
    root = f'{root}/neurofee'
os.chdir(root)
print('작업 폴더:', os.getcwd())
print(sorted(os.listdir()))
```

```python
import subprocess, os
os.environ.setdefault("NEUROFEE_OFFLINE", "1")
steps = [
    ("00 합성 데이터", "python scripts/00_make_synthetic.py", "data/synthetic/synthetic_clinical.csv"),
    ("01 라벨",       "python scripts/01_check_labels.py --synthetic", "data/processed/labels.csv"),
    ("02 특징",       "python scripts/02_extract_features.py --synthetic", "data/processed/features.csv"),
    ("03 학습",       "python scripts/03_train_difficulty.py", "models/difficulty_xgb.joblib"),
    ("04 에이전트",   "python scripts/04_run_agents.py --hardest", "reports"),
]
for name, cmd, out in steps:
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    ok = r.returncode == 0 and os.path.exists(out)
    print(("✅" if ok else "❌"), name, "→", out)
    if not ok:
        print(r.stderr[-1500:]); break
```
